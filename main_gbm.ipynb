```python
from tqdm.auto import tqdm
from data_loader import get_data
from evaluation import main_evaluation, testing
import debug 

from models.gbm import GBM

stocks = ['GAZP', 'SBER', 'GMKN', 'RTKM', 'MGNT']
```

### Нормальное распределение

#### Палитры параметров:
1. **`loc`** ($\mu$) — параметр локации (математическое ожидание / центр распределения).
2. **`scale`** ($\sigma$) — параметр масштаба (стандартное отклонение, $\sigma > 0$).

#### Функция плотности вероятности (PDF):
$$f(x) = \frac{1}{\sigma \sqrt{2\pi}} \exp\left( -\frac{(x - \mu)^2}{2\sigma^2} \right)$$

#### Функция распределения (CDF):
$$F(x) = \frac{1}{2} \left[ 1 + \text{erf}\left( \frac{x - \mu}{\sigma \sqrt{2}} \right) \right]$$
*где $\text{erf}(z)$ — функция ошибок.*

```python
model_gmm = GBM('normal')

with tqdm() as pbar:
    for st in stocks:
        tqdm.write(f"Stock: {st}")
        log_profit = get_data(st)
        
        I = main_evaluation(model_gmm, log_profit, pbar=pbar)
        
        testing(I)
        print()
```

### Распределение Лапласа

#### Палитры параметров:
1. **`loc`** ($\mu$) — параметр локации (медиана / центр распределения).
2. **`scale`** ($b$) — параметр масштаба ($b > 0$). Дисперсия равна $\sigma^2 = 2b^2$.

#### Функция плотности вероятности (PDF):
$$f(x) = \frac{1}{2b} \exp\left( -\frac{|x - \mu|}{b} \right)$$

#### Функция распределения (CDF):
$$F(x) = \begin{cases} 
\frac{1}{2} \exp\left( \frac{x - \mu}{b} \right), & \text{если } x < \mu \\[10pt]
1 - \frac{1}{2} \exp\left( -\frac{x - \mu}{b} \right), & \text{если } x \ge \mu 
\end{cases}$$

```python
model_gmm = GBM('laplace')

with tqdm() as pbar:
    for st in stocks:
        tqdm.write(f"Stock: {st}")
        log_profit = get_data(st)
        
        I = main_evaluation(model_gmm, log_profit, pbar=pbar)
        
        testing(I)
        print()
```

### Асимметричное нормальное распределение

#### Палитры параметров:
1. **`a`** ($\alpha$) — параметр асимметрии (форма). При $\alpha = 0$ переходит в обычное нормальное распределение.
2. **`loc`** ($\mu$) — параметр локации (сдвиг).
3. **`scale`** ($\omega$) — параметр масштаба ($\omega > 0$).

#### Функция плотности вероятности (PDF):
$$f(x) = \frac{2}{\omega} \phi\left(\frac{x - \mu}{\omega}\right) \Phi\left(\alpha \frac{x - \mu}{\omega}\right)$$
*где $\phi(z)$ и $\Phi(z)$ — PDF и CDF стандартного нормального распределения соответственно.*

```python
model_gmm = GBM('skew')

with tqdm() as pbar:
    for st in stocks:
        tqdm.write(f"Stock: {st}")
        log_profit = get_data(st)
        
        I = main_evaluation(model_gmm, log_profit, pbar=pbar)
        
        testing(I)
        print()
```

### Распределение Стьюдента

#### Палитры параметров:
1. **`df`** ($\nu$) — параметр формы (число степеней свободы, $\nu > 0$). Определяет тяжесть хвостов.
2. **`loc`** ($\mu$) — параметр локации (центр симметрии распределения).
3. **`scale`** ($\lambda$) — параметр масштаба ($\lambda > 0$). При $\nu > 2$ связан с дисперсией формулой $\sigma^2 = \lambda^2 \frac{\nu}{\nu-2}$.

#### Функция плотности вероятности (PDF):
$$f(x) = \frac{\Gamma\left(\frac{\nu+1}{2}\right)}{\Gamma\left(\frac{\nu}{2}\right) \sqrt{\pi \nu} \lambda} \left( 1 + \frac{1}{\nu} \left(\frac{x - \mu}{\lambda}\right)^2 \right)^{-\frac{\nu+1}{2}}$$
*где $\Gamma(z)$ — гамма-функция.*

```python
model_gmm = GBM('student')

with tqdm() as pbar:
    for st in stocks:
        tqdm.write(f"Stock: {st}")
        log_profit = get_data(st)
        
        I = main_evaluation(model_gmm, log_profit, pbar=pbar)
        
        testing(I)
        print()
```

### Обобщенное нормальное распределение

#### Палитры параметров:
1. **`beta`** ($\beta$) — параметр формы (крутость пика и хвостов, $\beta > 0$). При $\beta = 1$ это распределение Лапласа, при $\beta = 2$ — нормальное, при $\beta \to \infty$ стремится к равномерному.
2. **`loc`** ($\mu$) — параметр локации (математическое ожидание / центр).
3. **`scale`** ($\alpha$) — параметр масштаба ($\alpha > 0$).

#### Функция плотности вероятности (PDF):
$$f(x) = \frac{\beta}{2\alpha \Gamma(1/\beta)} \exp\left( -\left(\frac{|x - \mu|}{\alpha}\right)^\beta \right)$$

```python
model_gmm = GBM('gennorm')

with tqdm() as pbar:
    for st in stocks:
        tqdm.write(f"Stock: {st}")
        log_profit = get_data(st)
        
        I = main_evaluation(model_gmm, log_profit, pbar=pbar)
        
        testing(I)
        print()
```
